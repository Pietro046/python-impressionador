#### 8. Vamos criar um conversor de temperatura. Faça um Programa que peça a temperatura em graus Fahrenheit, transforme e mostre a temperatura em graus Celsius.
$C = \frac{5}{9}(F-32)$

```python
temperatura_fh = int(input("Digite a temperatura em fahrenheit: "))
temperatura_celsius = (temperatura_fh - 32) * 5 / 9
print(f"A temperatura convertida é {temperatura_celsius}")
```

#### 9. Faça um Programa que peça a temperatura em graus Celsius, transforme e mostre em graus Fahrenheit.
$F = \frac{9}{5}C + 32$

```python
temperatura_celsius = float(input("Digite a temperatura em celsius: "))
temperatura_fh = (temperatura_celsius * 9 / 5) + 32
print(f"A temperatura é {temperatura_fh}")
```

##### Calcule o desconto do sindicato (5% do salário bruto)

```python
salario_horas = float(input("Digite quanto ganha por hora: "))
horas_mensal = float(input("Digite quantas horas trabalha por mes: "))
salario_bruto = salario_horas * horas_mensal

desconto_sindicato = salario_bruto * 5 / 100
salario_liquido = salario_bruto - desconto_sindicato

print(f"Salario Bruto: R$ {salario_bruto:.2f}")
print(f"Desconto Sindicato: R$ {desconto_sindicato:.2f}")
print(f"Salario Liquido: R$ {salario_liquido:.2f}")
```

##### Calcule o salário líquido (salário bruto - descontos)

```python
salario_horas = float(input("Digite seu salário por hora: "))
horas_mensal = float(input("Digite quantas horas trabalhou por mes: "))
salario_bruto = salario_horas * horas_mensal
desconto_inss = salario_bruto * 8 / 100
desconto_sindicato = salario_bruto * 5 / 100
desconto_ir = salario_bruto * 11 / 100
salario_liquido = salario_bruto - desconto_ir - desconto_inss - desconto_sindicato

print(salario_liquido)
```
